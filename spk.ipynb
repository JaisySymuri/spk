```python
import pandas as pd

df = pd.read_csv("students.csv")

# Separate features & target
X = df.drop(columns=["exam_score", "student_id"])
y = df["exam_score"]

# One-hot encoding for categorical variables
X = pd.get_dummies(X, drop_first=True)
X = X.astype(float)   # 👈 ensures numeric
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

# Get weights
coefficients = pd.Series(model.coef_, index=X.columns)
coefficients = coefficients.sort_values(key=abs, ascending=False)

print(coefficients)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor()
rf.fit(X, y)

importance = pd.Series(rf.feature_importances_, index=X.columns)
importance = importance.sort_values(ascending=False)

print(importance)
```

```python
import shap

# 1. Sample data
sample_X = X.sample(500, random_state=42)

# 2. Fast explainer
explainer = shap.TreeExplainer(rf)

# 3. Compute SHAP
shap_values = explainer.shap_values(sample_X)

# 4. Plot
shap.summary_plot(shap_values, sample_X)
```

```python
rf = RandomForestRegressor(
    n_estimators=50,
    max_depth=10,
    n_jobs=-1,
    random_state=42
)

rf.fit(X, y)
```
